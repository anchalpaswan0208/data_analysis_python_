# Handling missing data


```python
#it will delete the duplicates of primary key
data.dropna(subset = ["Customer_ID"], inplace=True)
```

```python
#checking the data after deleting
data.shape
```

```python
#Checking the effect on each column
data.isnull().sum()
```

```python
#let's start fixing from the age column

data["Age"].unique()

#Note: From the age column , u can notice that it cobines a float value with strings and somewhere it is just a float value,
#we need to extract the int value from it

```

```python

# data['Age'].apply(lambda x : re.findall('[0-9]+',str(x)))
```

```python
import re
def extract_age(age):
    age_num = re.findall('[0-9]+', str(age))
    if len(age_num) > 0:
        return age_num[0]
    else:
        return np.nan

data["Age"] = data['Age'].apply(lambda x: extract_age(x))
```

```python
print(data['Age'].unique())
```

```python
df_age = data[data["Age"] != 'nan years']['Age']

```

```python
# age_median = df_age.dropna().astype('int64').median()
age_median = int(df_age.dropna().astype('int64').median())
```

```python
data.replace('nan years',age_median, inplace = True)
```

```python
# data['Age'] = pd.to_numeric(data['Age'], errors='coerce')
# data['Age'].fillna(age_median, inplace=True)
# data['Age'] = data['Age'].astype('Int64')
# data["Age"].unique()
```

```python
data['Age'] = pd.to_numeric(data['Age'], errors='coerce')
data['Age'].fillna(age_median, inplace=True)
data['Age'] = data['Age'].astype('Int64')
```
